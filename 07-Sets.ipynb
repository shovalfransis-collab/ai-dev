# Sets in Python!

**Sets** are an unordered collection of **unique** elements. Unlike lists or tuples, sets automatically remove duplicates and don't maintain order.

Think of a set like a bag of unique items:
- **No duplicates** - each element appears only once
- **No order** - elements have no specific position
- **Fast membership testing** - quickly check if something is in the set

## 1. Creating Sets

Sets are created using **curly braces** `{}` or the `set()` function.

### Important:
- Use `set()` to create an empty set (not `{}` which creates an empty dictionary)
- Elements must be **immutable** (strings, numbers, tuples)
- Elements are automatically **unique** - duplicates are removed

### Creating an Empty Set:

```python
# Create an empty set
x = set()
print("Empty set:", x)
print("Type:", type(x))
```

```python
# Create a set with elements directly
numbers = {1, 2, 3, 4, 5}
print("Numbers set:", numbers)

# Create a set with duplicate elements (duplicates are automatically removed)
duplicates = {1, 2, 2, 3, 3, 3, 4}
print("Set with duplicates removed:", duplicates)
```

```python
# Convert a list to a set (removes duplicates)
lst = [1, 1, 1, 2]
my_set = set[int](lst)
print("Original list:", lst)
print("Converted to set:", my_set)
```

```python
# Convert set back to list
back_to_list = list(my_set)
print("Back to list:", back_to_list)
print("Type:", type(back_to_list))
```

### Important: Sets vs Dictionaries

```python
# Important: Curly braces {} can mean dictionary or set
empty_dict = {}  # This creates an empty DICTIONARY
empty_set = set()  # This creates an empty SET

print("Empty dict:", empty_dict, "- Type:", type(empty_dict))
print("Empty set:", empty_set, "- Type:", type(empty_set))
```

```python
# But with elements, curly braces create a set
my_set = {1, 2, 3}
print("Set with elements:", my_set, "- Type:", type(my_set))
```

## 2. Adding Elements to Sets

Use the `.add()` method to add a single element to a set.

**Note:** Sets automatically ignore duplicate elements!

### Using .add() Method:

```python
# Create empty set
x = set()
print("Initial set:", x)

# Add elements
x.add(1)
x.add(2)
print("After adding 1 and 2:", x)
```

```python
# Try adding the same element multiple times
x.add(1)
x.add(1)
x.add(1)
x.add(1)
x.add(1)
print("After adding 1 multiple times:", x)
print("Notice: Only unique elements are kept!")
```

```python
# Add a different element
x.add(3)
print("After adding 3:", x)
```

```python
# Try adding an existing element again
x.add(1)
print("After trying to add 1 again:", x)
print("The set remains unchanged - no duplicates allowed!")
```

## 3. Practical Use: Removing Duplicates

**One of the most common uses of sets is to remove duplicates from a list!**

```python
# List with many duplicate elements
list1 = [1, 1, 2, 2, 3, 4, 5, 6, 1, 1]
print("Original list with duplicates:", list1)

# Convert to set to get unique values
unique_set = set(list1)
print("Converted to set (duplicates removed):", unique_set)
```

```python
# Convert back to list if needed
unique_list = list(unique_set)
print("Back to list (unique values only):", unique_list)
```

### Real-World Example:

```python
# Example: Finding unique visitors
visitors = ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Alice', 'David']
print("All visitors:", visitors)

unique_visitors = set(visitors)
print("Unique visitors:", unique_visitors)
print(f"Total visits: {len(visitors)}")
print(f"Unique visitors: {len(unique_visitors)}")
```

### Membership Testing (in):

```python
my_set = {1, 2, 3, 4, 5}

print("Is 3 in the set?", 3 in my_set)
print("Is 10 in the set?", 10 in my_set)
print("Is 10 NOT in the set?", 10 not in my_set)
```

## 4. Summary

### What we learned:

1. **Creating Sets:**
   - Use curly braces: `{1, 2, 3}`
   - Use `set()` for empty set or converting from list
   - Sets automatically remove duplicates

2. **Adding Elements:**
   - `.add(element)` - add single element
   - `.update(iterable)` - add multiple elements

3. **Removing Elements:**
   - `.remove(element)` - raises error if not found
   - `.discard(element)` - doesn't raise error if not found
   - `.clear()` - remove all elements

4. **Common Uses:**
   - Remove duplicates from lists
   - Check membership (very fast!)
   - Mathematical set operations

### Key Takeaway:
**Sets are perfect for storing unique elements and performing fast membership testing!**
