## 3. Reassignment and Updates

Variables can be **reassigned** to new values. The old value is replaced.

````python
# Initial assignment
a = 5
print("Initial value:", a)
````

````python
# Can even change the type!
a = "Hello"
print("Changed to string:", a)
````

````python
# Start with a = 10
a = 10
print("Starting value:", a)
````

````python
# Use a to redefine a (add a to itself)
a = a + a
print("After a = a + a:", a)  # 10 + 10 = 20
````

````python
# Do it again
a = a + a
print("After doing it again:", a)  # 20 + 20 = 40
````

### Shorthand Assignment Operators

Python provides shortcuts for updating variables:

````python
# Start fresh
a = 5
print("Starting value:", a)
````

````python
# a = a + 10 is the same as a += 10
a += 10  # Add 10 to a
print("After a += 10:", a)
````

````python
# More shorthand operators
a -= 5   # Subtract 5 (a = a - 5)
print("After a -= 5:", a)
````

````python
a *= 2   # Multiply by 2 (a = a * 2)
print("After a *= 2:", a)
````

````python
a /= 4   # Divide by 4 (a = a / 4)
print("After a /= 4:", a)
````

**Summary of Shorthand Operators:**

| Operator | Long Form | Shorthand |
|----------|-----------|------------|
| Addition | `a = a + 5` | `a += 5` |
| Subtraction | `a = a - 5` | `a -= 5` |
| Multiplication | `a = a * 5` | `a *= 5` |
| Division | `a = a / 5` | `a /= 5` |
| Floor Division | `a = a // 5` | `a //= 5` |
| Modulo | `a = a % 5` | `a %= 5` |
| Exponentiation | `a = a ** 5` | `a **= 5` |

### Assign Same Value to Multiple Variables:

````python
# All three variables get the same value
x = y = z = 0
print(x, y, z)
````

````python
# Assign multiple values at once
first_name, last_name = "John", "Doe"
print("First name:", first_name)
print("Last name:", last_name)
````

### 🔄 Swapping Variables

Python makes it easy to swap values between variables:

````python
# Swap the values in one line!
x, y = y, x
print("After swap: x =", x, ", y =", y)
````

````python
# Same variable, different types
my_var = 5           # Integer
print(type(my_var))

my_var = 5.5         # Float
print(type(my_var))

my_var = "Hello"     # String
print(type(my_var))
````

````python
# Check types of different variables
name = "Python"
age = 30
price = 19.99

print("Type of name:", type(name))
print("Type of age:", type(age))
print("Type of price:", type(price))
````

````python
# Get user's name
name = input("Enter your name: ")
print("Hello,", name)
````

### ⚠️ Input Returns Strings

Even if the user types a number, `input()` returns it as a string:

````python
# This looks like a number, but it's actually a string!
num = input("Enter a number: ")
print("You entered:", num)
print("Type:", type(num))
````

## 7. Type Conversion

Convert between different data types using:
- `int()` - Convert to integer
- `float()` - Convert to float
- `str()` - Convert to string

### Converting Strings to Numbers:

````python
# Convert string to integer
num_str = "42"
num_int = int(num_str)
print("String:", num_str, "Type:", type(num_str))
print("Integer:", num_int, "Type:", type(num_int))
````

````python
# Now we can do math!
result = int("10") + 5
print(result)
````

````python
# Convert string to float
price_str = "19.99"
price_float = float(price_str)
print("String:", price_str, "Type:", type(price_str))
print("Float:", price_float, "Type:", type(price_float))
````

### Converting Numbers to Strings:

````python
# Convert number to string
age = 25
age_str = str(age)
print("Number:", age, "Type:", type(age))
print("String:", age_str, "Type:", type(age_str))
````

````python
# Now we can concatenate with strings
name = "Alice"
age = 25
message = name + " is " + str(age) + " years old"
print(message)
````

### Using Input with Type Conversion:

````python
# Get a number from user and convert it
age = int(input("Enter your age: "))
next_year = age + 1
print("Next year you will be", next_year)
````

````python
# Calculator example
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
result = num1 + num2
print("Sum:", result)
````

## 8. String Formatting with Variables

There are several ways to include variables in strings:

### Method 1: String Concatenation

````python
# Using + to combine strings
first_name = "John"
last_name = "Doe"
message = "My name is " + first_name + " " + last_name
print(message)
````

### Method 2: F-Strings (Recommended!) 🌟

````python
# Using f-strings (Python 3.6+)
first_name = "John"
last_name = "Doe"
age = 30

# Put 'f' before the string and use {variable_name}
message = f"My name is {first_name} {last_name} and I am {age} years old"
print(message)
````

````python
# F-strings can include expressions
price = 19.99
quantity = 3
print(f"Total: ${price * quantity}")
````

### Method 3: .format() Method

````python
# Using .format() method
message = "My name is {} {} and I am {} years old".format(first_name, last_name, age)
print(message)
````

## 🎓 Summary

### What we learned:

1. **Variable Assignment:**
   - Use `=` to assign values: `variable_name = value`
   - Variables store data for later use

2. **Naming Rules:**
   - Must start with letter or underscore
   - No spaces (use `_` instead)
   - Only letters, numbers, and underscores
   - Use lowercase with underscores (snake_case)
   - Avoid Python keywords

3. **Reassignment:**
   - Variables can be updated: `a = a + 5`
   - Shorthand operators: `+=`, `-=`, `*=`, `/=`

4. **Multiple Assignment:**
   - Same value: `x = y = z = 0`
   - Different values: `x, y = 1, 2`
   - Swapping: `x, y = y, x`

5. **Dynamic Typing:**
   - Variables can change type
   - Use `type()` to check type

6. **User Input:**
   - `input()` always returns a string
   - Convert with `int()`, `float()`, or `str()`

7. **String Formatting:**
   - F-strings: `f"Hello {name}"`
   - Concatenation: `"Hello " + name`
   - .format(): `"Hello {}".format(name)`
