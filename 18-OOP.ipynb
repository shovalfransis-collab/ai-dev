### `self`

`self` refers to the instance. You write it in the method definition, but Python passes it automatically when you call the method.

```python
# Example showing self clearly
class Person:
    def __init__(self, name):
        self.name = name
        print(f"self is: {self}")
        print(f"self.name is: {self.name}")

person1 = Person("Alice")
print(f"person1 is: {person1}")
print(f"person1.name is: {person1.name}")
```

```python
# Methods can accept additional parameters
class Dog:
    """A dog class with parameterized methods."""
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def birthday(self):
        """Increment the dog's age."""
        self.age += 1
        print(f"Happy Birthday {self.name}! Now {self.age} years old.")
    
    def fetch(self, item):
        """Simulate fetching an item."""
        print(f"{self.name} fetched the {item}!")
    
    def describe(self):
        """Return a description of the dog."""
        return f"{self.name} is {self.age} years old."

# Test the methods
my_dog = Dog("Buddy", 3)
print(my_dog.describe())
my_dog.fetch("ball")
my_dog.fetch("stick")
my_dog.birthday()
print(my_dog.describe())
```

```python
class LibraryBook:
    """A class representing a library book."""

    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_checked_out = False


    def describe_book(self):
        """Return a description of the book."""
        if self.is_checked_out:
            status = "Checked out"
        else:
            status = "Available"
        return f"'{self.title}' by {self.author} ({self.year}) - {status}"
   

    def check_out(self):
        """Mark the book as checked out."""
        self.is_checked_out = True

    def return_book(self):
        """Mark the book as returned."""
        self.is_checked_out = False
        print(f"The book '{self.title}' has been returned.")

# Example usage:
book1 = LibraryBook("Harry Potter", "J.K. Rowling", 1997)
print(book1.describe_book())   # Show details before checkout

book1.check_out()
print(book1.describe_book())   # Should show as checked out

book1.return_book()
print(book1.describe_book())   # Should show as available again
```

```python
# Create multiple independent objects
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def sit(self):
        print(f"{self.name} is now sitting.")

# Create three different dogs
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)
dog3 = Dog("Bella", 2)

# Each dog is independent
dog1.sit()
dog2.sit()
dog3.sit()

print(f"\n{dog1.name} is {dog1.age} years old.")
print(f"{dog2.name} is {dog2.age} years old.")
print(f"{dog3.name} is {dog3.age} years old.")
```

## Part 2: Working with Attributes

Default values, and validation logic inside methods.

```python
# Class with default values
class Car:
    """A simple car model."""
    
    def __init__(self, make, model, year):
        """Initialize car attributes."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0  # Default value
    
    def get_descriptive_name(self):
        """Return a formatted description."""
        return f"{self.year} {self.make} {self.model}"
    
    def read_odometer(self):
        """Print the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

# Create a car - odometer starts at 0
my_car = Car("Ford", "Mustang", 2020)
print(my_car.get_descriptive_name())
my_car.read_odometer()

# Modify the odometer
my_car.odometer_reading = 100
my_car.read_odometer()
```

```python
# Validation in methods
class Car:
    """A car with validation."""
    
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0
    
    def read_odometer(self):
        print(f"This car has {self.odometer_reading} miles on it.")
    
    def update_odometer(self, mileage):
        """Set odometer - prevent rollback."""
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")
    
    def increment_odometer(self, miles):
        """Add miles - prevent negative values."""
        if miles >= 0:
            self.odometer_reading += miles
        else:
            print("You can't add negative miles!")

# Test validation
my_car = Car("Honda", "Civic", 2021)
my_car.update_odometer(100)
my_car.read_odometer()

# Try to roll back - should fail
my_car.update_odometer(50)
my_car.read_odometer()

# Valid increment
my_car.increment_odometer(25)
my_car.read_odometer()

# Invalid increment
my_car.increment_odometer(-10)
my_car.read_odometer()
```

## Part 3: Advanced OOP

**Class attributes** are shared by all instances; **instance attributes** are unique per object.

```python
# Comparing class and instance attributes
class Dog:
    """A dog class demonstrating attribute types."""
    
    # Class attribute - shared by all dogs
    species = "Canis familiaris"
    
    def __init__(self, name, age):
        # Instance attributes - unique to each dog
        self.name = name
        self.age = age

# Create multiple dogs
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)

# Instance attributes are different
print(f"{dog1.name} is {dog1.age} years old.")
print(f"{dog2.name} is {dog2.age} years old.")

# Class attribute is the same for all
print(f"\n{dog1.name}'s species: {dog1.species}")
print(f"{dog2.name}'s species: {dog2.species}")

# Can also access via class name
print(f"\nAll dogs are: {Dog.species}")
```

### Inheritance

A child class inherits attributes and methods from a parent class.

```python
# Basic inheritance example
class Animal:
    """A general animal class (parent)."""
    
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def make_sound(self):
        print(f"{self.name} makes a sound.")
    
    def eat(self):
        print(f"{self.name} is eating.")

# Child class inherits from Animal
class Dog(Animal):
    """A dog class (child of Animal)."""
    pass  # Inherits everything from Animal

# Create a dog - has all Animal methods
my_dog = Dog("Buddy", 3)
print(f"Name: {my_dog.name}")
print(f"Age: {my_dog.age}")
my_dog.make_sound()
my_dog.eat()
```

```python
# Child class with extra attributes
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Dog(Animal):
    """Dog with breed attribute."""
    
    def __init__(self, name, age, breed):
        # Initialize parent attributes
        super().__init__(name, age)
        # Add dog-specific attribute
        self.breed = breed
    
    def describe(self):
        return f"{self.name} is a {self.age}-year-old {self.breed}."

my_dog = Dog("Buddy", 3, "Golden Retriever")
print(my_dog.describe())
```

### Method Overriding

A child class can replace a parent method by redefining it.

```python
# Method overriding example
class Animal:
    """Parent class."""
    
    def __init__(self, name):
        self.name = name
    
    def make_sound(self):
        print(f"{self.name} makes a generic sound.")

class Dog(Animal):
    """Dog with overridden make_sound."""
    
    def make_sound(self):
        print(f"{self.name} says: Woof! Woof!")

class Cat(Animal):
    """Cat with overridden make_sound."""
    
    def make_sound(self):
        print(f"{self.name} says: Meow!")

class Cow(Animal):
    """Cow with overridden make_sound."""
    
    def make_sound(self):
        print(f"{self.name} says: Moo!")

# Create animals
animals = [
    Dog("Buddy"),
    Cat("Whiskers"),
    Cow("Bessie")
]

# Each makes its own sound
for animal in animals:
    animal.make_sound()
```

### `super()`

`super()` calls the parent class's methods (commonly `__init__`).

```python
# Complete inheritance example
class Car:
    """A regular car (parent class)."""
    
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0
    
    def get_descriptive_name(self):
        return f"{self.year} {self.make} {self.model}"
    
    def read_odometer(self):
        print(f"This car has {self.odometer_reading} miles on it.")
    
    def update_odometer(self, mileage):
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

class ElectricCar(Car):
    """An electric car (child class)."""
    
    def __init__(self, make, model, year, battery_size=75):
        # Initialize parent class
        super().__init__(make, model, year)
        # Add electric-specific attribute
        self.battery_size = battery_size
        self.battery_level = 100
    
    def describe_battery(self):
        """Describe the battery."""
        print(f"This car has a {self.battery_size}-kWh battery.")
        print(f"Battery level: {self.battery_level}%")
    
    def charge(self):
        """Charge the battery."""
        self.battery_level = 100
        print("Battery fully charged!")
    
    def drive(self, miles):
        """Simulate driving."""
        battery_used = miles * 0.3  # 0.3% per mile
        if self.battery_level >= battery_used:
            self.battery_level -= battery_used
            self.odometer_reading += miles
            print(f"Drove {miles} miles. Battery: {self.battery_level:.1f}%")
        else:
            print("Not enough battery! Please charge.")

# Test the electric car
my_tesla = ElectricCar("Tesla", "Model S", 2023, battery_size=100)
print(my_tesla.get_descriptive_name())

my_tesla.describe_battery()
my_tesla.drive(50)
my_tesla.read_odometer()

my_tesla.drive(200)
my_tesla.charge()
my_tesla.describe_battery()
```
