# if, elif, and else Statements in Python!

**Conditional statements** allow your code to make decisions and perform different actions based on different conditions.

Think of it like this:
- **if** - "If this condition is true, do this"
- **elif** - "Else, if this other condition is true, do that"
- **else** - "Otherwise, if none of the above are true, do this"

These statements are fundamental for:
- **Making decisions** in your programs
- **Controlling program flow** based on conditions
- **Responding to user input** differently
- **Validating data** before processing

## 1. The if Statement

The `if` statement executes code **only if** a condition is True.

### Syntax:
```python
if condition:
    # code to execute if condition is True
```

**Important:** Note the colon `:` and the **indentation** (4 spaces or 1 tab)

### Basic if Examples:

````python
if 2>1:
    print("BOKER TOV")

print("Ori")
````

````python
# Simple if statement
if False:
    print('It was true!')
````

````python
password = input("enter your pass:")

if password.lower() == "python4ever":
    print("great")
else:
    print("try again")

````

````python
num = 6

if num % 2 == 0:
    print("even")
else:
    print("odd")
````

````python
num = int(input("enter a number"))



if num % 3 == 0:
    print("mod0")
elif num % 3 == 1:
    print("mod1")
else:
    print("mod2")
````

````python
# if with comparison
if 3 > 2:
    print('3 is greater than 2!')
````

````python
# Another comparison
if 3 < 5:
    print('3 is less than 5!')
````

````python
# If condition is False, nothing happens
if 3 > 5:
    print('This will NOT print because 3 is not greater than 5')
    
print('This line always prints - it\'s not indented!')
````

### Using Variables:

````python
# Using a variable in if statement
x = True

if x:
    print('x was True!')
````

````python
# Practical example: checking age
age = 20

if age >= 18:
    print('You are an adult!')
````

````python
# Checking string equality
name = 'Alice'

if name == 'Alice':
    print('Hello Alice!')
````

## 2. The else Statement

The `else` statement executes code when the `if` condition is **False**.

### Syntax:
```python
if condition:
    # code if condition is True
else:
    # code if condition is False
```

````python
# if-else example
x = False

if x:
    print('x was True!')
else:
    print('x was False!')
````

````python
# Another example
x = True

if x:
    print('x was True!')
else:
    print('I will be printed in any case where x is not true')
````

### Practical Examples:

````python
# Checking if a number is even or odd
number = 7

if number % 2 == 0:
    print(f'{number} is even')
else:
    print(f'{number} is odd')
````

````python
# Checking age for voting
age = 16

if age >= 18:
    print('You can vote!')
else:
    print('You cannot vote yet')
````

````python
# Password check
password = 'secret123'
user_input = 'secret456'

if password == user_input:
    print('Access granted!')
else:
    print('Access denied - incorrect password')
````

## 3. The elif Statement

The `elif` (else if) statement allows you to check **multiple conditions**.

### Syntax:
```python
if condition1:
    # code if condition1 is True
elif condition2:
    # code if condition1 is False and condition2 is True
else:
    # code if all conditions are False
```

**Note:** You can have as many `elif` statements as you need!

````python
# Simple elif example
person = 'Sammy'

if person == 'Sammy':
    print('Welcome Sammy!')
else:
    print("Welcome, what's your name?")
````

````python
# Adding elif
person = 'George'

if person == 'Sammy':
    print('Welcome Sammy!')
elif person == 'George':
    print('Welcome George!')
else:
    print("Welcome, what's your name?")
````

---
## 4. Multiple Conditions with elif

You can chain as many `elif` statements as needed to handle multiple cases.

````python
# Multiple elif statements
loc = 'Bank'

if loc == 'Auto Shop':
    print('Welcome to the Auto Shop!')
elif loc == 'Bank':
    print('Welcome to the bank!')
elif loc == 'Store':
    print('Welcome to the store!')
else:
    print('Where are you?')
````

````python
# Grade classification
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print(f'Score: {score}, Grade: {grade}')
````

````python
# Age categories
age = 25

if age < 13:
    category = 'Child'
elif age < 20:
    category = 'Teenager'
elif age < 65:
    category = 'Adult'
else:
    category = 'Senior'

print(f'Age: {age}, Category: {category}')
````

## 5. Nested if Statements

You can put `if` statements inside other `if` statements - this is called **nesting**.

````python
# Nested if example
age = 20
has_id = True

if age >= 18:
    print('You are old enough')
    if has_id:
        print('You can enter - you have ID')
    else:
        print('You need to show ID')
else:
    print('You are too young')
````

````python
# Another nested example
num = 15

if num > 0:
    print('Number is positive')
    if num % 2 == 0:
        print('And it is even')
    else:
        print('And it is odd')
elif num < 0:
    print('Number is negative')
else:
    print('Number is zero')
````

## 6. Indentation Rules

**Indentation is CRITICAL in Python!**

- Use **4 spaces** or **1 tab** for each level of indentation
- All code at the same indentation level belongs together
- Incorrect indentation will cause errors!

````python
# Correct indentation
x = True

if x:
    print('This is indented correctly')  # 4 spaces
    print('This is also part of the if block')  # 4 spaces
    
print('This is NOT part of the if block')  # No indentation
````

````python
# Multiple levels of indentation
x = 10

if x > 5:  # Level 0
    print('x is greater than 5')  # Level 1 (4 spaces)
    if x > 8:  # Level 1 (4 spaces)
        print('x is also greater than 8')  # Level 2 (8 spaces)
        print('Still level 2')  # Level 2 (8 spaces)
    print('Back to level 1')  # Level 1 (4 spaces)
    
print('Back to level 0')  # Level 0 (no spaces)
````

## 7. Practical Examples

Let's put it all together with real-world examples!

### Example 1: Simple Calculator

````python
# Simple calculator
num1 = 10
num2 = 5
operation = '+'

if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    result = num1 / num2
else:
    result = 'Invalid operation'

print(f'{num1} {operation} {num2} = {result}')
````

### Example 2: Temperature Classifier

````python
# Temperature classifier
temperature = 25

if temperature > 30:
    print('It\'s hot outside!')
elif temperature > 20:
    print('It\'s warm outside')
elif temperature > 10:
    print('It\'s cool outside')
else:
    print('It\'s cold outside!')

print(f'Temperature: {temperature}°C')
````

### Example 3: Login System

````python
# Simple login system
username = 'admin'
password = 'secret123'

user_input_name = 'admin'
user_input_pass = 'secret123'

if user_input_name == username:
    if user_input_pass == password:
        print('Login successful! Welcome!')
    else:
        print('Incorrect password')
else:
    print('Username not found')
````

### Example 4: Discount Calculator

````python
# Discount based on purchase amount
purchase_amount = 150

if purchase_amount >= 200:
    discount = 0.20  # 20% discount
    discount_label = '20%'
elif purchase_amount >= 100:
    discount = 0.10  # 10% discount
    discount_label = '10%'
elif purchase_amount >= 50:
    discount = 0.05  # 5% discount
    discount_label = '5%'
else:
    discount = 0  # No discount
    discount_label = '0%'

final_price = purchase_amount * (1 - discount)

print(f'Purchase amount: ${purchase_amount}')
print(f'Discount: {discount_label}')
print(f'Final price: ${final_price}')
````

1. **if Statement:**

2. **else Statement:**

   - Can have multiple elif statements
   - Syntax: `elif condition:`

   ```python
   if condition1:

       # code
   elif condition3:
       # code
   else:
       # code
   ```

5. **Indentation:**
   - Use 4 spaces or 1 tab
   - Critical for Python syntax
   - Defines code blocks

6. **Nested Statements:**
   - Can put if statements inside other if statements
   - Each level adds 4 more spaces
