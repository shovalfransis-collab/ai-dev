## 1. Types of Numbers in Python

Python supports several types of numbers. We'll focus on the two main types:

### 🔷 Integers
- Whole numbers without a decimal point
- Can be positive or negative
- Examples: `2`, `-5`, `1000`, `0`

### 🔷 Floating Point Numbers (Floats)
- Numbers with a decimal point
- Or numbers in exponential notation
- Examples: `2.0`, `-0.5`, `3.14`, `2e2` (200), `3E-2` (0.03)

### Summary Table:

| Examples | Number Type | Explanation |
|---------|-----------|-------|
| 1, 2, -5, 1000 | Integer | Numbers without a decimal point |
| 1.2, -0.5, 2e2, 3E2 | Float | Numbers with a decimal point or scientific notation |

## 2. Division - Two Important Types!

### ➗ Regular Division - `/`
Always returns a float

````python
# Regular division
7 / 3
````

````python
# Even when the division is exact, we get a float
10 / 2
````

### ⬇️ Floor Division - `//`
Returns only the integer part of the division result (no rounding!)

````python
# Floor division
7 // 4  # 1.75 -> 1 (drops the decimal part)
````

````python
# Another example
10 // 3  # 3.333... -> 3
````

````python
# Comparison between the two division types
print("Regular division:", 17 / 5)
print("Floor division:", 17 // 5)
````

### 🔄 Modulo - `%`
Returns the remainder of the division

````python
# Remainder of division
7 % 2  # 7 divided by 2 = 3 with remainder 1
````

````python
# Another example
17 % 5  # 17 divided by 5 = 3 with remainder 2
````

````python
# Common use: checking if a number is even or odd
print(10 % 2)  # 0 - even
print(11 % 2)  # 1 - odd
````

### 🎯 Relationship between Floor Division and Modulo:

```
17 = (17 // 5) * 5 + (17 % 5)
17 = 3 * 5 + 2
17 = 15 + 2 ✓
```

## 4. Order of Operations

Python follows the standard mathematical order of operations:

1. **Parentheses** - `()`
2. **Exponents** - `**`
3. **Multiplication & Division** - `*`, `/`, `//`, `%`
4. **Addition & Subtraction** - `+`, `-`

💡 Remember: **PEMDAS** (Parentheses, Exponents, Multiplication/Division, Addition/Subtraction)

````python
# Multiplication before addition
2 + 10 * 10 + 3
````

````python
# Using parentheses to change order of operations
(2 + 10) * (10 + 3)
````

````python
# More complex example
2 ** 3 + 10 / 2 - 3
````

````python
# With parentheses
(2 ** 3 + 10) / (2 - 3)
````
