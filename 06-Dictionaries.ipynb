# Dictionaries in Python! 🗂️

We've been learning about **sequences** in Python (strings, lists, tuples), but now we're going to learn about **mappings**.

A **dictionary** is a mapping that stores data as **key-value pairs**. Instead of accessing data by position (like in lists), you access data by its **key**.

Think of a real dictionary:
- **Word** (key) → **Definition** (value)
- **Name** (key) → **Phone number** (value)
- **Username** (key) → **Password** (value)

## 1. Creating Dictionaries

Dictionaries are created using **curly braces** `{}` with **key:value** pairs separated by **commas**.

### Syntax:
```python
my_dict = {key1: value1, key2: value2, key3: value3}
```

**Important:**
- Keys must be **immutable** (strings, numbers, tuples)
- Keys must be **unique** (no duplicates)
- Values can be **any type** and can be duplicated

### Basic Dictionary Creation:

````python
person = {"first_name": "jon", "children": ["Odelle", "Batya"]}

person["last_name"] = "zouari"
person
````

````python
# Simple dictionary
person = {'name': 'John', 'age': 30, 'city': 'New York'}
print(person)
print(type(person))
````

````python
# Dictionary with different value types
student = {
    'name': 'Alice',
    'age': 20,
    'grades': [85, 90, 92],
    'is_graduated': False
}
print(student)
````

````python
# Using variables as values
name = 'Bob'
age = 25

user = {'username': name, 'user_age': age}
print(user)
````

### Empty Dictionary:

````python
# Create an empty dictionary
empty_dict = {}
print("Empty dict:", empty_dict)
print("Type:", type(empty_dict))

# Alternative using dict()
empty_dict2 = dict()
print("Empty dict2:", empty_dict2)
````

### Using dict() Constructor:

````python
# Creating dictionary using dict()
person = dict(name='Sarah', age=28, city='Boston')
print(person)
````

## 2. Accessing Values

Access dictionary values using **square brackets** `[]` with the key, or using the `.get()` method.

````python
person = {'name': 'John', 'age': 30, 'city': 'New York'}

# Access by key
print("Name:", person['name'])
print("Age:", person['age'])
print("City:", person['city'])
````

````python
# ❌ KeyError if key doesn't exist
# Uncomment to see the error:
# print(person['phone'])  # KeyError: 'phone'

print("Use .get() to avoid KeyError!")
````

````python
person = {'name': 'John', 'age': 30, 'city': 'New York'}

# Get existing key
print("Name:", person.get('name'))

# Get non-existing key (returns None)
print("Phone:", person.get('phone'))

# Get with default value
print("Phone:", person.get('phone', 'Not available'))
````

### Adding New Items:

````python
# Start with empty dictionary
person = {}
print("Initial:", person)

# Add items
person['name'] = 'Alice'
person['age'] = 25
person['city'] = 'Boston'

print("After adding:", person)
````

### Modifying Existing Items:

````python
person = {'name': 'John', 'age': 30, 'city': 'New York'}
print("Original:", person)

# Modify value
person['age'] = 31
person['city'] = 'Los Angeles'

print("Modified:", person)
````

### Updating with Operations:

````python
scores = {'math': 85, 'english': 90}
print("Original scores:", scores)

# Update score
scores['math'] = scores['math'] + 5
print("After update:", scores)

# Using shorthand
scores['english'] += 3
print("After shorthand:", scores)
````

### Deleting Items:

````python
# Delete using del
person = {'name': 'John', 'age': 30, 'city': 'New York', 'temp': 'delete me'}
print("Before:", person)

del person['temp']
print("After del:", person)
````

````python
# Delete using .pop() (returns the value)
person = {'name': 'John', 'age': 30, 'city': 'New York'}

removed_age = person.pop('age')
print("Removed:", removed_age)
print("Dictionary after pop:", person)
````

### .keys() - Get All Keys:

````python
keys = person.keys()
print("Keys:", keys)
print("Type:", type(keys))

# Convert to list
keys_list = list(keys)
print("As list:", keys_list)
````

### .values() - Get All Values:

````python
values = person.values()
print("Values:", values)

# Convert to list
values_list = list(values)
print("As list:", values_list)
````

### .items() - Get All Key-Value Pairs:

````python
items = person.items()
print("Items:", items)

# Convert to list of tuples
items_list = list(items)
print("As list:", items_list)
````

### .update() - Merge Dictionaries:

````python
person = {'name': 'John', 'age': 30}
print("Original:", person)

# Add multiple items at once
person.update({'city': 'New York', 'job': 'Engineer'})
print("After update:", person)

# Update existing and add new
person.update({'age': 31, 'country': 'USA'})
print("After second update:", person)
````

### .clear() - Remove All Items:

````python
temp_dict = {'a': 1, 'b': 2, 'c': 3}
print("Before clear:", temp_dict)

temp_dict.clear()
print("After clear:", temp_dict)
````

### .copy() - Create a Shallow Copy:

````python
original = {'name': 'John', 'age': 30}
copied = original.copy()

# Modify copy
copied['age'] = 25

print("Original:", original)
print("Copied:", copied)
````

### .setdefault() - Get or Set Default:

````python
person = {'name': 'John', 'age': 30}

# Get existing key
name = person.setdefault('name', 'Unknown')
print("Name:", name)

# Get non-existing key (adds it with default)
city = person.setdefault('city', 'New York')
print("City:", city)
print("Dictionary after setdefault:", person)
````

## 5. Nested Dictionaries

Dictionaries can contain other dictionaries, creating complex data structures.

### Simple Nested Dictionary:

````python
# Dictionary within dictionary
person = {
    'name': 'John',
    'age': 30,
    'address': {
        'street': '123 Main St',
        'city': 'New York',
        'zip': '10001'
    }
}

print("Full person:", person)
print("\nAddress:", person['address'])
print("City:", person['address']['city'])
print("Street:", person['address']['street'])
````

### Multiple Levels of Nesting:

````python
# Deeply nested dictionary
company = {
    'name': 'TechCorp',
    'departments': {
        'engineering': {
            'manager': 'Alice',
            'employees': ['Bob', 'Charlie', 'David']
        },
        'sales': {
            'manager': 'Eve',
            'employees': ['Frank', 'Grace']
        }
    }
}

# Access nested values
print("Engineering manager:", company['departments']['engineering']['manager'])
print("First sales employee:", company['departments']['sales']['employees'][0])
````

### Dictionary of Dictionaries:

````python
# Multiple people
people = {
    'person1': {'name': 'Alice', 'age': 25},
    'person2': {'name': 'Bob', 'age': 30},
    'person3': {'name': 'Charlie', 'age': 35}
}

print("Person1:", people['person1'])
print("Person2's name:", people['person2']['name'])
print("Person3's age:", people['person3']['age'])
````

## 6. Dictionary Operations

### Checking if Key Exists (in):

````python
person = {'name': 'John', 'age': 30, 'city': 'New York'}

# Check if key exists
print("'name' in person:", 'name' in person)
print("'phone' in person:", 'phone' in person)
print("'phone' not in person:", 'phone' not in person)
````

````python
# Using in with if statement
if 'age' in person:
    print(f"Age: {person['age']}")
else:
    print("Age not found")
````

### Dictionary Length:

````python
person = {'name': 'John', 'age': 30, 'city': 'New York'}
print("Number of key-value pairs:", len(person))
````

### Merging Dictionaries (Python 3.9+):

````python
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}

# Using | operator (Python 3.9+)
merged = dict1 | dict2
print("Merged:", merged)

# Original dictionaries unchanged
print("dict1:", dict1)
print("dict2:", dict2)
````

````python
# Alternative: unpacking method (works in all versions)
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}

merged = {**dict1, **dict2}
print("Merged with unpacking:", merged)
````

## 🎓 Summary

### What we learned:

1. **Creating Dictionaries:**
   - Use curly braces: `{key: value}`
   - Keys must be immutable and unique
   - Values can be any type

2. **Accessing Values:**
   - Square brackets: `dict[key]`
   - `.get()` method: safer, returns None or default if key missing

3. **Modifying:**
   - Add: `dict[new_key] = value`
   - Modify: `dict[key] = new_value`
   - Delete: `del dict[key]` or `dict.pop(key)`

4. **Dictionary Methods:**
   - `.keys()` - get all keys
   - `.values()` - get all values
   - `.items()` - get key-value pairs
   - `.update()` - merge dictionaries
   - `.get()` - safe access
   - `.pop()` - remove and return
   - `.clear()` - remove all items

5. **Nested Dictionaries:**
   - Dictionaries within dictionaries
   - Access with multiple brackets: `dict[key1][key2]`

6. **Operations:**
   - Membership: `key in dict`
   - Length: `len(dict)`
   - Merge: `dict1 | dict2` or `{**dict1, **dict2}`


### Key Takeaway:
**Dictionaries store data as key-value pairs** - perfect for representing real-world objects and relationships!
