# `*args` and `**kwargs`

`*args` lets a function accept any number of **positional** arguments (collected as a tuple). Without it, you'd need a fixed parameter for each value:

```python
def myfunc(a=0, b=0, c=0, d=0, e=0):
    return sum((a, b, c, d, e)) * .05

myfunc(40, 60)
```

## `*args`

A parameter prefixed with `*` accepts an arbitrary number of arguments as a tuple. The name (`args`) is just convention.

```python
def myfunc(*args):
    return sum(args) * .05

myfunc(1, 2, 3, 4, 5, 6, 7, 832,12,3,3,234,23,5,23,5,325)
```

## `**kwargs`

`**kwargs` collects arbitrary **keyword** arguments into a dictionary, which you can loop over with `.items()`.

```python
def agents(**kwargs):
    return kwargs

agent1 = agents(role="data_analyst", goal="analyze big data")
print(agent1)
```

```python
def get_info(**kwargs):
    return kwargs


username1 = input("enter your username")
passw = input("enter your password")

user = get_info(username=username1,passw=passw)
```

```python
user
```

```python
# kwargs is a normal dict - access values by key
agent1["goal"]
```

### Looping over `**kwargs`

```python
def pass_database(**kwargs):
    for user, passw in kwargs.items():
        print(f"username -> {user}, password -> {passw}")

pass_database(jonathan="python4ever", nathan="agents4ever")
```

### Checking for a specific keyword

```python
def myfunc(**kwargs):
    if 'fruit' in kwargs:
        print('My favorite fruit is {}'.format(kwargs['fruit']))
    else:
        print("I don't like fruit")

myfunc(fruit='apple')
myfunc(vegetable='carrot')
```
