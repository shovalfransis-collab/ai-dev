# while Loops in Python!

**while loops** repeatedly execute code as long as a condition remains True.

Think of a while loop as saying:
- "**While** this condition is true, keep doing this"

### Key Differences from for Loops:
- **for loop**: "Do this for each item in a sequence" (known number of iterations)
- **while loop**: "Keep doing this until a condition becomes False" (unknown number of iterations)

### Syntax:
```python
while condition:
    # code to execute while condition is True
```

⚠️ **Warning:** Be careful not to create infinite loops!

## 1. Basic while Loops

A while loop continues executing as long as the condition is True.

````python
while 2 > 1:
    print("hi")
````

````python
counter = 0

while counter < 5:
    counter = counter + 1 # counter += 1
    print(counter)
    
````

````python
counter = 0

while counter != 5:
    print(counter)
    counter = counter + 1  # counter += 1
````

````python
# More detail with an f-string
x = 0

while x < 10:
    print(f'x is currently: {x}')
    x += 1  # x = x + 1
````

````python
for num in [1,2,3,4]:

    if num == 3:
        pass

    print(num)
````

````python
for num in [1,2,3,4]:
    pass
````

### Comparison: if vs while

**if** executes once, **while** executes repeatedly!

````python
# Using if - executes ONCE
count = 0

if count < 5:
    count += 1
    print(f'if statement: count = {count}')
````

````python
# Using while - executes MULTIPLE TIMES
count = 0

while count < 5:
    count += 1
    print(f'while loop: count = {count}')
````

## 2. while with else

You can add an `else` clause that executes when the loop completes normally (not via break).

````python
x = 0

while x < 10:
    print(f'x is currently: {x}')
    x += 1
else:
    print('All Done!')
````

---
## 3. break Statement

The `break` statement **exits the loop immediately**, even if the condition is still True.

**Important:** When using `break`, the `else` block will NOT execute!

````python
count = 0

while count < 10:
    count += 1
    print(count)

    if count == 5:
        print('Breaking at 5!')
        break
````

---
## 4. continue Statement

The `continue` statement **skips the rest of the current iteration** and goes back to check the condition.

````python
# Print only even numbers
num = 0

while num < 10:
    num += 1

    # Skip odd numbers
    if num % 2 != 0:
        continue

    print(f'{num} is even')
````

### Difference: break vs continue

````python
# Using break - stops at 5
print('Using break:')
count = 0
while count < 10:
    count += 1
    if count == 5:
        break
    print(count)
````

````python
# Using continue - skips 5 but keeps going
print('Using continue:')
count = 0
while count < 10:
    count += 1
    if count == 5:
        continue
    print(count)
````

---
## 5. pass Statement

The `pass` statement does **nothing** - it's a placeholder.

Use it when:
- You're writing code structure but haven't filled in details yet
- Syntax requires a statement but you don't want to do anything

````python
x = 0

while x < 5:
    x += 1
    pass  # Placeholder - does nothing

print(f'Final x: {x}')
````

## 6. Practical Examples

Let's see real-world uses of while loops!

### Example 1: Processing a List Until Empty

````python
names = ['Jon', 'Ben', 'Dan']

print('Original list:', names)
print('\nProcessing:')

while names:  # While list is not empty
    name = names.pop()  # Remove and return last item
    print(f'  Processing: {name}')

print('\nFinal list:', names)
````

### Example 2: Finding a Value

````python
# Find first number divisible by both 7 and 13
num = 1
found = False

while not found:
    if num % 7 == 0 and num % 13 == 0:
        print(f'Found: {num} is divisible by both 7 and 13')
        found = True
    num += 1
````

### Example 3: Countdown Timer

````python
countdown = 10

while countdown > 0:
    print(countdown)
    countdown -= 1

print('Blast off!')
````

## 7. Common Patterns

### Pattern 1: Flag Pattern

````python
running = True
count = 0

while running:
    count += 1
    print(f'Running... {count}')

    if count >= 5:
        running = False

print('Stopped')
````

### Pattern 2: Sentinel Pattern (while + break + else)

````python
numbers = [1, 5, 3, 8, 2, 9, 4]
index = 0

print('Searching for number greater than 7...')

while index < len(numbers):
    if numbers[index] > 7:
        print(f'Found: {numbers[index]} at index {index}')
        break
    index += 1
else:
    print('Not found')
````

## 8. Infinite Loops (and How to Avoid Them)

⚠️ **Warning:** An infinite loop runs forever and can freeze your program!

````python
# ⚠️ DO NOT RUN THIS CODE! ⚠️  This is an infinite loop:
# while True:
#     print("I'm stuck in an infinite loop!")

# Common mistakes that cause infinite loops:
# 1) Forgetting to update the counter:
#    count = 0
#    while count < 10:
#        print(count)      # forgot count += 1  ->  INFINITE LOOP
#
# 2) Condition that never becomes False:
#    x = 10
#    while x > 5:
#        x += 1            # x keeps growing  ->  INFINITE LOOP

print('(Infinite loop examples are commented out for safety)')
````

### How to Avoid Infinite Loops:

1. **Always update** your loop variable
2. **Make sure** the condition can become False
3. **Use a counter** or flag as a safety measure
4. **Test carefully** with small numbers first

````python
# ✅ Safe while loop with a counter
count = 0
max_iterations = 10  # Safety limit

while count < max_iterations:
    print(f'Iteration {count}')
    count += 1  # Don't forget this!

print('Loop completed safely')
````

### What we learned:

1. **while Loop Syntax:**
   ```python
   while condition:
       # code
   ```

2. **while with else:** runs when the loop ends normally (no break).

3. **Control Statements:**
   - **break**: Exit the loop immediately

   - **while loop**: When you don't know how many iterations (until condition is False)

   - Always update your loop variable
   - Ensure the condition can become False
   - Use safety counters
   - Test with small values first

### Key Takeaway:
**while loops are perfect for repeating code until a condition is met when you don't know in advance how many times it will repeat. Just be careful to avoid infinite loops!**
