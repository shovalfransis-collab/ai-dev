# Strings in Python! 📝

Strings are used in Python to record text information, such as names, messages, or any textual data. Strings in Python are actually a **sequence**, which means Python keeps track of every element in the string in a specific order.

This allows us to use **indexing** to grab particular letters (like the first letter, or the last letter).

## 1. Creating Strings

To create a string in Python, you can use either:
- **Single quotes** `'`
- **Double quotes** `"`
- **Triple quotes** `'''` or `"""` (for multi-line strings)

### Basic String Creation:

````python
"jonathan"
````

````python
"I'm Jonathan"
````

````python
'He said "hello"'
````

````python
'''jonathan
zouari'''
````

````python
# Single word using single quotes
'hello'
````

````python
# Phrase using double quotes
"Hello World"
````

````python
# Both single and double quotes work the same
'This is also a string'
````

### 💡 Mixing Quotes

You can use double quotes inside single quotes, or vice versa:

````python
"I'm Jonathan"
````

````python
'he said "hello"'
````

````python
# Using double quotes inside single quotes
'He said "Hello"'
````

````python
'''my name is
jonathan'''
````

````python
# Using single quote (apostrophe) inside double quotes
"I'm learning Python"
````

````python
# ❌ This will cause an error - mismatched quotes!
# 'I'm learning Python'  # The apostrophe ends the string early
````

### Multi-line Strings

Use triple quotes for strings that span multiple lines:

````python
# Multi-line string with triple quotes
'''This is a multi-line string.
It can span several lines.
Very useful for long text!'''
````

## 2. Printing Strings

While Jupyter notebooks can display strings directly, the proper way to output strings is using the `print()` function.

### Direct Output vs Print:

````python
# If you have multiple strings, only the last one is shown
'Hello World 1'
'Hello World 2'  # Only this one appears
````

````python
print("jonathan","zouari","age")
````

### Special Characters in Strings

Python supports special escape characters:
- `\n` - New line
- `\t` - Tab
- `\\` - Backslash
- `\'` - Single quote
- `\"` - Double quote

````python
# Using \n for new line
print("My \nHello\tWorld")
````

````python
# Multiple new lines
print("First line")
print("\n")  # Empty line
print("Second line")
````

````python
# Count characters in a string
len('Hello World')
````

````python
# Spaces count as characters
len('Python')
````

````python
# Empty string has length 0
len('')
````

## 4. String Indexing

Strings are **sequences**, which means each character has a position (index). In Python:
- **Indexing starts at 0** (first character is at index 0)
- We use **square brackets** `[]` to access characters
- **Negative indexing** starts from the end (-1 is the last character)

### Visual Example:
```
String:  H  e  l  l  o
Index:   0  1  2  3  4
Negative:-5 -4 -3 -2 -1
```

````python
# First character (index 0)
'Hello World'[0]
````

````python
# Space is also a character (index 5)
'Hello World'[5]
````

````python
# Last character
'Hello World'[-1]
````

````python
# Third from the end
'Hello World'[-3]
````

## 5. String Slicing

**Slicing** allows you to grab a substring (portion of a string).

### Syntax:
```python
string[start:stop:step]
```

- **start**: Where to begin (inclusive)
- **stop**: Where to end (exclusive - does NOT include this index)
- **step**: How many characters to skip (default is 1)

### Important: The stop index is NOT included!

### Basic Slicing:

````python
# From index 1 to the end
'Hello World'[1:]
````

````python
# From beginning up to (but not including) index 3
'Hello World'[:3]
````

````python
# Everything (complete copy)
'Hello World'[:]
````

````python
# From index 2 to index 7 (not including 7)
'Hello World'[2:7]
````

### Slicing with Step:

````python
# Every character (step of 1 - default)
'Hello World'[::1]
````

````python
# Every other character (step of 2)
'Hello World'[::2]
````

````python
# Every third character
'Hello World'[::3]
````

### 🔄 Reversing a String

A step of -1 reverses the string:

````python
# Reverse the string
'Hello World'[::-1]
````

````python
# Reverse with step of 2
'Hello World'[::-2]
````

## 6. String Properties - Immutability

Strings in Python are **immutable**, which means:
- Once created, you **cannot change** individual characters
- You must create a new string instead

### Example of Immutability:

````python
# We can read a character
'Hello'[0]
````

````python
# ❌ But we CANNOT change it - this will cause an error!
# 'Hello'[0] = 'J'  # TypeError: 'str' object does not support item assignment
````

### ✅ What You CAN Do: Concatenation

You can create new strings by combining (concatenating) them:

````python
# Concatenate strings with +
'Hello' + ' ' + 'World'
````

````python
# Create a new string instead of modifying
'J' + 'ello'
````

````python
# Multiple concatenations
'Python' + ' ' + 'is' + ' ' + 'awesome!'
````

### String Repetition

Use the `*` operator to repeat strings:

````python
# Repeat a string 3 times
'Hello' * 3
````

````python
# Create a separator line
print('=' * 50)
````

````python
# Combining repetition with concatenation
'Ha' * 3 + '!'
````

## 7. Built-in String Methods

Strings have many built-in **methods** (functions that belong to the string object).

### Method Syntax:
```python
string.method()
```

**Important**: Methods don't change the original string (immutability), they return a new string!

### Case Conversion Methods:

````python
# Convert to uppercase
'hello world'.upper()
````

````python
# Convert to lowercase
'HELLO WORLD'.lower()
````

````python
# Title case (first letter of each word capitalized)
'hello world'.title()
````

````python
# Capitalize only the first letter
'hello world'.capitalize()
````

````python
# Swap case (upper becomes lower, lower becomes upper)
'Hello World'.swapcase()
````

### Splitting and Joining:

````python
# Split by spaces (returns a list)
'Hello World Python'.split()
````

````python
# Split by a specific character
'apple,banana,orange'.split(',')
````

````python
# Split by newline
'Line 1\nLine 2\nLine 3'.split('\n')
````

### Searching in Strings:

````python
# Check if string starts with a substring
'Hello World'.startswith('Hello')
````

````python
# Check if string ends with a substring
'Hello World'.endswith('World')
````

````python
# Find the index of a substring (-1 if not found)
'Hello World'.find('World')
````

````python
# Count occurrences of a substring
'Hello Hello Hello'.count('Hello')
````

### Replacing in Strings:

````python
# Replace all occurrences
'Hello World'.replace('World', 'Python')
````

````python
# Replace with limit (replace only first 2 occurrences)
'Hello Hello Hello'.replace('Hello', 'Hi', 2)
````

### Checking String Content:

````python
# Check if all characters are letters
'Hello'.isalpha()
````

````python
# Check if all characters are digits
'12345'.isdigit()
````

````python
# Check if all characters are alphanumeric (letters or digits)
'Hello123'.isalnum()
````

````python
# Check if all characters are lowercase
'hello'.islower()
````

````python
# Check if all characters are uppercase
'HELLO'.isupper()
````

### Whitespace Handling:

````python
# Remove leading and trailing whitespace
'   Hello World   '.strip()
````

````python
# Remove leading whitespace only
'   Hello World   '.lstrip()
````

````python
# Remove trailing whitespace only
'   Hello World   '.rstrip()
````

## 🎓 Summary

### What we learned:

1. **String Creation:**
   - Single quotes `'...'` or double quotes `"..."`
   - Triple quotes for multi-line strings `'''...'''`
   - Mixing quotes for apostrophes and quotation marks

2. **String Output:**
   - Use `print()` for proper output
   - Special characters: `\n` (newline), `\t` (tab)

3. **String Length:**
   - `len(string)` returns the number of characters

4. **Indexing:**
   - Positive: `string[0]` (first), `string[1]` (second)
   - Negative: `string[-1]` (last), `string[-2]` (second to last)

5. **Slicing:**
   - `string[start:stop:step]`
   - `string[::-1]` reverses the string

6. **Immutability:**
   - Cannot change individual characters
   - Create new strings with concatenation `+` or repetition `*`

7. **String Methods:**
   - Case: `.upper()`, `.lower()`, `.title()`, `.capitalize()`
   - Searching: `.find()`, `.count()`, `.startswith()`, `.endswith()`
   - Modifying: `.replace()`, `.split()`, `.strip()`
   - Checking: `.isalpha()`, `.isdigit()`, `.isalnum()`
