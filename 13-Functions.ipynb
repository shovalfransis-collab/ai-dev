# Functions in Python!

**Functions** are reusable blocks of code that perform specific tasks.

Think of a function as saying:
- "**Define** this task once, **use** it many times"

### Why Use Functions?
- **Avoid repetition**: Write code once, use it many times
- **Organization**: Break complex problems into smaller pieces
- **Maintainability**: Update code in one place
- **Reusability**: Use the same function across different projects

### Basic Syntax:
```python
def function_name(parameters):
    '''
    Docstring: Description of what the function does
    '''
    # Code to execute
    return result
```

## 1. Creating Basic Functions (def keyword)

We use the `def` keyword to **define** a function in Python.

### Simple Function with No Parameters

````python
def say_hello():
    print("Hello!")
````

````python
say_hello()
````

````python
def say_hello2(name: str) -> str:
    """
    Return a greeting for the given name.

    Args:
        name (str): The name of the person to greet.

    Returns:
        str: Greeting string.
    """
    return "hi " + name
````

````python
say_hello2("jonathan")
````

### Important: Functions Need Parentheses () to Be Called

````python
# Without parentheses - shows it's a function object
print(say_hello)

# With parentheses - actually calls the function
say_hello()
````

## 2. Function Parameters and Arguments

Functions can accept **parameters** (inputs) to make them more flexible and reusable.

### Function with One Parameter

````python
# Function that accepts a name parameter
def greeting(name):
    print(f'Hello {name}')

greeting('Anat')
````

### Function with Multiple Parameters

````python
def print2(first_name, last_name):
    print(f"Hi {first_name} {last_name}")

print2("Jonathan", "Zouari")
````

### Using Named Arguments (Keyword Arguments)

````python
# You can specify arguments by name
def info_person(first_name, last_name):
    print(f"Hello {first_name} {last_name}")

# Order doesn't matter when using keyword arguments
info_person(last_name="Smith", first_name="John")
````

---
## 3. Return vs Print

**CRITICAL CONCEPT**: `return` and `print()` are NOT the same!

- **print()**: Displays output but doesn't save it
- **return**: Sends a value back that can be saved and used later

### Function with print()

````python
def add_print(a, b):
    print(a + b)

# Try to save the result
result = add_print(10, 5)
print(f"Result: {result}")
print(f"Type: {type(result)}")
````

### Function with return

````python
def add_return(a, b):
    return a + b

# Save and use the result
result = add_return(10, 5)
print(f"Result: {result}")
print(f"Type: {type(result)}")
print(f"Result + 1 = {result + 1}")
````

## 4. Default Parameters

You can set **default values** for parameters. If no argument is provided, the default is used.

### Example: Default Last Name

````python
def introduce(first_name: str = "John", last_name: str = "Doe") -> None:
    """
    Print the first and last name of a person.

    Args:
        first_name (str, optional): The first name of the person. Defaults to "John".
        last_name (str, optional): The last name of the person. Defaults to "Doe".
    """
    print(f"my first name is {first_name}, and my last name is {last_name}")
````

````python
introduce()
````

````python
def info_person(first_name, last_name="Smith"):
    print(f"Hello {first_name} {last_name}")

# Call with both arguments
info_person("John", "Doe")

# Call with only first name - uses default last name
info_person("Jon")
````

### Example: Default Exponent

````python
def power(a, b=2):
    return a ** b

# With both arguments
print(f"2^5 = {power(2, 5)}")

# With default b=2
print(f"3^2 = {power(3)}")

# Using keyword arguments
print(f"2^5 = {power(b=5, a=2)}")
````

### Example: Type Hints (Modern Python)

````python
def greet(name: str, age: int = 0) -> str:
    """
    Returns a greeting string including the person's name and age.

    Parameters:
        name (str): The name of the person.
        age (int): The age of the person. Defaults to 0.

    Returns:
        str: A greeting message.
    """
    return f"Hello {name}, you are {age} years old."

print(greet("Jonathan", 30))
print(greet("Sarah"))
````

## 5. Adding Logic Inside Functions

Functions can contain if/else statements, loops, and complex logic.

### Example 1: Check if Number is Even

````python
def even_check(number):
    return number % 2 == 0

print(f"Is 20 even? {even_check(20)}")
print(f"Is 21 even? {even_check(21)}")
````

### Example 2: Check if a List Contains Even Numbers

````python
# Returns True as soon as ANY even number is found
def check_even_list(num_list):
    for number in num_list:
        if number % 2 == 0:
            return True   # Found one -> stop and return
    return False          # Finished loop with no even number

print(check_even_list([1, 2, 3]))  # True (2 is even)
print(check_even_list([1, 3, 5]))  # False
````

### Example 3: Return All Even Numbers from a List

````python
def get_even_numbers(num_list):
    even_numbers = []

    for number in num_list:
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers

print(get_even_numbers([1, 2, 3, 4, 5, 6]))  # [2, 4, 6]
print(get_even_numbers([1, 3, 5]))            # []
````

## 6. Multiple Return Values

Functions can return multiple values using **tuples**.

### Example 1: Tuple Unpacking Basics

````python
# We can loop through tuples and unpack them
stock_prices = [('AAPL', 200), ('GOOG', 300), ('MSFT', 400)]

for stock, price in stock_prices:
    print(f"{stock}: ${price}")
````

### Example 2: Employee of the Month

````python
# Function that returns both name and hours
def employee_of_month(work_hours):
    """
    Returns the employee with the most hours worked.

    Args:
        work_hours: List of tuples (name, hours)

    Returns:
        Tuple of (name, hours) for the top employee
    """
    max_hours = 0
    employee_name = ''

    for name, hours in work_hours:
        if hours > max_hours:
            max_hours = hours
            employee_name = name

    return (employee_name, max_hours)

work_hours = [('Abby', 100), ('Billy', 400), ('Cassie', 800), ('Shula', 300)]
name, hours = employee_of_month(work_hours)
print(f"Employee of the month: {name} with {hours} hours")
````

## 7. Practical Examples

Let's see some real-world function examples!

### Example 1: Temperature Converter

````python
def celsius_to_fahrenheit(celsius):
    """
    Converts temperature from Celsius to Fahrenheit.

    Args:
        celsius (float): Temperature in Celsius

    Returns:
        float: Temperature in Fahrenheit
    """
    return (9/5) * celsius + 32

print(f"0°C = {celsius_to_fahrenheit(0)}°F")
print(f"100°C = {celsius_to_fahrenheit(100)}°F")
````

### Example 2: Is Prime Number

````python
def is_prime(num):
    """
    Check if a number is prime.

    Args:
        num (int): Number to check

    Returns:
        bool: True if prime, False otherwise
    """
    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True

print(f"Is 7 prime? {is_prime(7)}")
print(f"Is 10 prime? {is_prime(10)}")
````

````python
from random import shuffle

def shuffle_list(mylist):
    """Shuffles the list and returns it."""
    shuffle(mylist)
    return mylist

# Example: 'O' ends up in a random position
example = [' ', 'O', ' ']
print(shuffle_list(example))
````

## 9. Summary

### What we learned:

1. **Function Syntax:**
   ```python
   def function_name(parameters):
       '''Docstring'''
       # Code here
       return result
   ```

2. **Key Concepts:**
   - Use `def` keyword to define functions
   - Call functions with parentheses `()`
   - Functions can accept parameters (inputs)
   - Use `return` to send values back (not `print`!)
   - Default parameters provide fallback values

3. **Return vs Print:**
   - **print()**: Displays output, returns None
   - **return**: Sends value back to be saved/used
   - Always use `return` when you need to save results!

4. **Functions Can:**
   - Contain if/else logic
   - Use loops (for, while)
   - Return multiple values (tuples)
   - Call other functions
   - Have default parameters

5. **Best Practices:**
   - Use descriptive function names
   - Add docstrings to explain what functions do
   - Keep functions focused on one task
   - Use type hints for clarity (modern Python)

### Key Takeaway:
**Functions are the building blocks of larger programs. They help you write DRY (Don't Repeat Yourself) code that's easier to read, test, and maintain.**
