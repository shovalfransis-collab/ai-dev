# List Comprehensions in Python!

**List comprehensions** provide a concise way to create lists in Python.

Think of a list comprehension as saying:
- "Create a list by doing **this** to each item in **that**"

### Key Differences from Traditional Loops:
- **Traditional for loop**: Multiple lines, explicit append statements
- **List comprehension**: One line, more readable and Pythonic

### Basic Syntax:
```python
[expression for item in iterable]
```

### With Condition:
```python
[expression for item in iterable if condition]
```

### Benefits:
- More concise and readable
- Often faster than traditional loops
- More Pythonic code style

## 1. Basic List Comprehensions

List comprehensions allow you to create lists in a single, readable line.

### Simple Example: Squaring Numbers

Goal: turn `[1, 2, 3, 4]` into `[1, 4, 9, 16]` (square each number).

````python
lst = []

for num in [1,2,3,4]:
    lst.append(num**2)
````

````python
lst
````

````python
[num**2 for num in [1,2,3,4]]
````

````python
# Create a list of squares
squares = [x**2 for x in range(1, 6)]
squares
````

### Example: First Letter of Each Name

````python
names = ["jon", "dan", "beni"]

[name[0] for name in names]
````

### Example: Letters from a String

````python
# Extract every letter from a string into a list
letters = [letter for letter in 'word']
print(letters)
````

---
## 2. Comparison: Traditional Loop vs List Comprehension

Let's see the difference between traditional for loops and list comprehensions!

````python
# Traditional for loop - Multiple lines
lst = []

for i in range(11):
    lst.append(i**2)

print('Traditional loop result:', lst)
````

````python
# List comprehension - One line!
lst = [i**2 for i in range(11)]

print('List comprehension result:', lst)
````

---
## 3. List Comprehensions with Conditions

You can add an **if statement** to filter items in your list comprehension.

### Example 1: Filter Even Numbers

````python
# Get only even numbers from range
even_numbers = [x for x in range(0, 12) if x % 2 == 0]

print('Even numbers:', even_numbers)
````

### Example 2: Filter Odd Numbers and Square Them

````python
# Get odd numbers and square them
odd_squares = [x**2 for x in range(10) if x % 2 != 0]

print('Odd numbers squared:', odd_squares)
````

---
## 4. List Comprehensions with Multiple Conditions

You can combine multiple conditions using `and` or `or`.

### Example 1: Numbers Divisible by Both 2 and 3

````python
# Numbers divisible by both 2 AND 3
divisible_by_both = [x for x in range(30) if x % 2 == 0 and x % 3 == 0]

print('Divisible by 2 and 3:', divisible_by_both)
````

### Example 2: Filter Words by Length

````python
# Get words with length greater than 3
words = ['cat', 'dog', 'elephant', 'bird', 'lion', 'tiger']
long_words = [word for word in words if len(word) > 3]

print('Words longer than 3 characters:', long_words)
````

---
## 5. Nested List Comprehensions

List comprehensions can be nested, but use them carefully - they can become hard to read!

### Example: Flattening a Matrix

````python
# Flatten a 2D list (matrix) into a 1D list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Nested list comprehension to flatten
flattened = [num for row in matrix for num in row]

print('Matrix:', matrix)
print('Flattened:', flattened)
````

## 6. Practical Examples

Let's see real-world uses of list comprehensions!

### Example 1: Temperature Conversion (Celsius to Fahrenheit)

````python
# Convert Celsius to Fahrenheit
celsius = [0, 10, 20.1, 34.5]

# Formula: F = (9/5) * C + 32
fahrenheit = [((9/5) * temp + 32) for temp in celsius]

print('Celsius:', celsius)
print('Fahrenheit:', fahrenheit)
````

### Example 2: String Manipulation - Uppercase

````python
# Convert list of names to uppercase
names = ['alice', 'bob', 'charlie', 'david']

uppercase_names = [name.upper() for name in names]

print('Original:', names)
print('Uppercase:', uppercase_names)
````

### Example 3: Filter Positive Numbers

````python
# Filter out negative numbers
numbers = [-5, 10, -3, 8, 0, -1, 7]

positive_only = [num for num in numbers if num > 0]

print('Original numbers:', numbers)
print('Positive only:', positive_only)
````

## 7. When NOT to Use List Comprehensions

While list comprehensions are powerful, they're not always the best choice.

### Avoid When:

1. **Too Complex**: If it becomes hard to read, use a traditional loop
2. **Side Effects**: List comprehensions should create lists, not perform actions
3. **Multiple Operations**: Better to use a function with a loop
4. **Debugging**: Traditional loops are easier to debug

````python
# BAD: Too complex and hard to read
complex_bad = [x**2 if x % 2 == 0 else x**3 for x in range(10) if x % 3 != 0]

print('Complex (hard to understand):', complex_bad)
````

````python
# GOOD: Clear and readable with a traditional loop
result = []

for x in range(10):
    if x % 3 != 0:        # Skip multiples of 3
        if x % 2 == 0:    # Even numbers
            result.append(x**2)
        else:             # Odd numbers
            result.append(x**3)

print('Traditional loop (easier to understand):', result)
````

   - When debugging is needed

### Key Takeaway:
