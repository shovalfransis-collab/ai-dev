# Comparison Operators in Python!

**Comparison operators** allow us to compare values and return a **Boolean result** (True or False).

These operators are fundamental for:
- **Making decisions** in if statements
- **Controlling loops** with while statements
- **Filtering data** based on conditions
- **Comparing values** in your programs

---
## 1. What are Comparison Operators?

Comparison operators compare two values and return either **True** or **False**.

### Table of Comparison Operators:

| Operator | Description | Example (a=3, b=4) | Result |
|----------|-------------|-------------------|--------|
| `==` | Equal to | `a == b` | False |
| `!=` | Not equal to | `a != b` | True |
| `>` | Greater than | `a > b` | False |
| `<` | Less than | `a < b` | True |
| `>=` | Greater than or equal to | `a >= b` | False |
| `<=` | Less than or equal to | `a <= b` | True |

---
## 2. Equal to (==)

The `==` operator checks if two values are **equal**.

**Important:** Don't confuse `==` (comparison) with `=` (assignment)!
- `=` assigns a value to a variable
- `==` compares two values

```python
2 > 1
```

```python
2 == 1
```

```python
a = 5
b = 7

a != b
```

```python
1 not in [1,2,3]
```

```python
# Equal comparison
print("2 == 2:", 2 == 2)
print("1 == 0:", 1 == 0)
```

```python
# Comparing variables
a = 3
b = 4
print(f"a = {a}, b = {b}")
print(f"a == b: {a == b}")
print(f"a == 3: {a == 3}")
```

```python
# Comparing strings
print("'hello' == 'hello':", 'hello' == 'hello')
print("'hello' == 'Hello':", 'hello' == 'Hello')  # Case sensitive!
```

### Important Note: == vs =

**Common mistake:**

```python
# = is assignment (gives a value to a variable)
x = 5
print("x =", x)

# == is comparison (checks if values are equal)
print("x == 5:", x == 5)
print("x == 10:", x == 10)
```

---
## 3. Not Equal to (!=)

The `!=` operator checks if two values are **not equal**.

```python
# Not equal comparison
print("2 != 1:", 2 != 1)
print("2 != 2:", 2 != 2)
```

```python
# Practical example
password = "secret123"
user_input = "secret456"

print(f"Password: {password}")
print(f"User input: {user_input}")
print(f"Passwords don't match: {password != user_input}")
```

---
## 4. Greater Than (>)

The `>` operator checks if the left value is **greater than** the right value.

```python
# Greater than comparison
print("2 > 1:", 2 > 1)
print("2 > 4:", 2 > 4)
print("2 > 2:", 2 > 2)  # Equal values return False
```

```python
# Practical example
score = 85
passing_grade = 60

print(f"Score: {score}")
print(f"Passing grade: {passing_grade}")
print(f"Passed: {score > passing_grade}")
```

---
## 5. Less Than (<)

The `<` operator checks if the left value is **less than** the right value.

```python
# Less than comparison
print("2 < 4:", 2 < 4)
print("2 < 1:", 2 < 1)
print("2 < 2:", 2 < 2)  # Equal values return False
```

```python
# Practical example
age = 15
minimum_age = 18

print(f"Age: {age}")
print(f"Minimum age: {minimum_age}")
print(f"Too young: {age < minimum_age}")
```

---
## 6. Greater Than or Equal (>=)

The `>=` operator checks if the left value is **greater than or equal to** the right value.

```python
# Greater than or equal comparison
print("2 >= 2:", 2 >= 2)  # Equal values return True
print("2 >= 1:", 2 >= 1)
print("2 >= 4:", 2 >= 4)
```

```python
# Practical example
balance = 100
minimum_balance = 100

print(f"Balance: {balance}")
print(f"Minimum balance: {minimum_balance}")
print(f"Balance is sufficient: {balance >= minimum_balance}")
```

---
## 7. Less Than or Equal (<=)

The `<=` operator checks if the left value is **less than or equal to** the right value.

```python
# Less than or equal comparison
print("2 <= 2:", 2 <= 2)  # Equal values return True
print("2 <= 4:", 2 <= 4)
print("2 <= 1:", 2 <= 1)
```

```python
# Practical example
items_in_cart = 10
max_items = 10

print(f"Items in cart: {items_in_cart}")
print(f"Maximum items allowed: {max_items}")
print(f"Cart is within limit: {items_in_cart <= max_items}")
```

## 8. Comparing Different Data Types

### Comparing Strings:

```python
# Strings are compared alphabetically
print("'a' < 'b':", 'a' < 'b')
print("'apple' < 'banana':", 'apple' < 'banana')
print("'A' < 'a':", 'A' < 'a')  # Uppercase comes before lowercase in ASCII
```

### Comparing Lists:

```python
# Lists are compared element by element
print("[1, 2, 3] == [1, 2, 3]:", [1, 2, 3] == [1, 2, 3])
print("[1, 2, 3] == [1, 2, 4]:", [1, 2, 3] == [1, 2, 4])
print("[1, 2] < [1, 3]:", [1, 2] < [1, 3])
```

### Comparing Booleans:

```python
# True is treated as 1, False as 0
print("True == 1:", True == 1)
print("False == 0:", False == 0)
print("True > False:", True > False)
```

---
## 9. Chaining Comparisons

Python allows you to chain comparison operators for more readable code!

```python
# Check if a value is within a range
x = 5

# Traditional way
print("Traditional: (x > 1) and (x < 10):", (x > 1) and (x < 10))

# Python's chained comparison (cleaner!)
print("Chained: 1 < x < 10:", 1 < x < 10)
```

```python
# More examples of chained comparisons
age = 25
print(f"Age: {age}")
print("Is age between 18 and 65?", 18 <= age <= 65)

score = 95
print(f"\nScore: {score}")
print("Is score between 90 and 100?", 90 <= score <= 100)
```

```python
# Even longer chains are possible!
print("1 < 2 < 3 < 4:", 1 < 2 < 3 < 4)
print("1 < 2 < 3 > 2:", 1 < 2 < 3 > 2)
```

   - `>` - Greater than

   - Python allows: `1 < x < 10` instead of `(x > 1) and (x < 10)`
   - Makes code more readable!

### Key Takeaway:
**Comparison operators are essential for making decisions in your code - they return True or False based on comparing values!**
