# Nested Statements and Scope

Every variable name has a **scope** that determines where it's visible. Python resolves names using the **LEGB** rule:

- **L**ocal - names assigned inside the current function
- **E**nclosing - names in any enclosing function
- **G**lobal - names assigned at the top level of the module
- **B**uilt-in - names Python provides (e.g. `print`, `len`)

### Global variables
A name defined at the top level is global - visible in other cells and inside functions (for reading).

```python
name = 'This is a global name'
```

```python
print(name)
```

### Local variables
Variables assigned inside a function are local - changing them does not affect a same-named variable outside.

```python
x = 50

def func(x):
    print('x is', x)
    x = 2
    print('Changed local x to', x)
    return x


x = func(x)
print('x is still', x)
```

Inside `func`, the first `print` uses the parameter `x` that was passed in. Assigning `x = 2` creates a **local** name, so the global `x` outside the function is unaffected.

### Mutable arguments are different
You can't reassign a global from inside a function without `global`, but you *can* mutate a mutable object (like a list) that was passed in - because you're changing the object, not rebinding the name.

```python
lst = [1, 2, 3]

def change(some_list):
    some_list.pop()   # mutates the same list object

change(lst)
print(lst)   # [1, 2] - the original list was changed
```

### The `global` statement
To *assign* to a global variable from inside a function, declare it with `global` first. Without it, the assignment would create a new local name instead.

You can declare several at once: `global x, y, z`.

```python
count = 0

def increment():
    global count
    count += 1

increment()
increment()
print('count is now', count)   # 2 - the global was changed
```

### The `nonlocal` statement
`nonlocal` works like `global`, but for a variable in an **enclosing** function rather than the module's global scope.

```python
def outer():
    message = 'local to outer'

    def inner():
        nonlocal message
        message = 'changed by inner'

    inner()
    print(message)   # 'changed by inner'

outer()
```

### Summary

- Names resolve by the **LEGB** rule: Local -> Enclosing -> Global -> Built-in.
- Assigning inside a function creates a **local** name by default.
- Mutating a passed-in mutable object *does* affect the caller.
- Use `global` to rebind a module-level name, `nonlocal` to rebind an enclosing one.
- Reach for these sparingly - passing values in and returning results out is clearer.
