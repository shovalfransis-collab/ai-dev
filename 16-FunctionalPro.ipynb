# Functional Programming

**Functional programming** describes *what* you want computed rather than spelling out *how* to loop and mutate. Python supports it with tools like `lambda`, `map`, `filter`, and `zip`.

## Topics we'll cover:
1. Imperative vs Functional Approach
2. Lambda Expressions
3. `map()`
4. `filter()`
5. `zip()`
6. Recap: Unpacking Arguments with `*`

---
## 1. Imperative vs Functional Approach

The **imperative** style builds the result step by step with an explicit loop. The **functional** style applies a transformation to the whole sequence at once.

```python
numbers = [1, 2, 3, 4]

# Imperative Approach
result = []
for num in numbers:
    result.append(num * 3)
print("Imperative Approach:", result)  # [3, 6, 9, 12]
```

```python
# Functional Approach
result = list(map(lambda x: x * 3, numbers))
print("Functional Approach:", result)  # [3, 6, 9, 12]
```

---
## 2. Lambda Expressions

A `lambda` is a small anonymous function written in a single expression. `lambda x: x ** 2` is equivalent to a `def` that returns `x ** 2`.

```python
square = lambda x: x ** 2
print("Lambda Square of 6:", square(6))  # 36
```

---
## 3. Using `map()`

`map(func, iterable)` applies `func` to every item and returns a lazy map object, so we wrap it in `list()` to see the values.

```python
cubed = list(map(lambda x: x ** 3, numbers))
print("Cubed Numbers:", cubed)  # [1, 8, 27, 64]
```

---
## 4. Using `filter()`

`filter(func, iterable)` keeps only the items for which `func` returns `True`.

```python
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print("Even Numbers:", even_numbers)  # [2, 4]
```

---
## 5. Using `zip()`

`zip()` pairs up items from several iterables position by position.

```python
list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']
zipped = list(zip(list1, list2))
print("Zipped Pairs:", zipped)  # [(1, 'a'), (2, 'b'), (3, 'c')]
```

---
## 6. Recap: Unpacking Arguments with `*`

The `*` operator unpacks a sequence into separate positional arguments - handy when your values are already in a list.

```python
def add(a, b, c):
    return a + b + c

values = [2, 3, 4]
print(add(*values))  # 9 - same as add(2, 3, 4)
```

---
## Summary

- **Imperative** = describe *how* (explicit loops); **functional** = describe *what*.
- `lambda` writes a small anonymous function inline.
- `map` transforms every item, `filter` keeps matching items, `zip` pairs iterables.
- `map`/`filter`/`zip` are lazy - wrap them in `list()` to materialize the result.
- `*` unpacks a sequence into positional arguments.
